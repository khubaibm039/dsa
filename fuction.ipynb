```python
def my_func(*nums):
    total = 0
    for i in nums:
        total += i
    return total
print(my_func(1,2,3,4))
print(my_func(10, 20, 30,40, 50))
print(my_func(5))
```

```python
def my_function(*numbers):
  if len(numbers) == 0:
    return None
  max_num = numbers[0]
  for num in numbers:
    if num > max_num:
      max_num = num
  return max_num

print(my_function(3, 7, 2, 9, 1))
```

```python
def count_up_to(n):
    count = 1
    while count <= n:
        yield  count
        count += 1
for num in count_up_to(5):
    print(num)
```

```python
def large_scale(n):
    for i in range(n):
        yield i
gen = large_scale(1000000)
print(next(gen))
```

```python
def fibonacci():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b
gen = fibonacci()
for _ in range(10):
    print(next(gen))
```

```python
def echo_generator():
    while True:
        received = yield
        print("Received:", received)
gen = echo_generator()
next(gen)
gen.send("Hello")
gen.send('World')
```

```python
def my_gen():
    try:
        yield 1
        yield 2
        yield 3
    finally:
        print("Generator Closed")
gen = my_gen()
print(next(gen))
print(next(gen))
gen.close()
```
