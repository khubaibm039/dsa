```python
file = open('sample.txt', 'r')
content = file.readlines()
content = list(map(str.strip, content))
num_line = len(content)
print(num_line)
num_word = 0
for line in range(len(content)):
    l = (content[line]).split()
    num_word += len(l)
print(num_word)
num_char = 0
for line in content:
    l = list(map(str.split, line))
    filter_l = list(filter(lambda x : len(x)> 0 , l))
    num_char += len(filter_l)
print(num_char)
counter_of_string = f"Number of line = {num_line}\nNumber of words = {num_word} \nNumber of character = {num_char}\n"

with open ('counter_of_string.txt', 'w') as file:
    file.write(counter_of_string)
```

```python
from functools import reduce

with open ('sample.txt', 'r') as file:
    strings = file.readlines()
    print(len(strings))

    number_of_words = list(map(lambda x: len(x.split()), strings))
    total_number_of_words = reduce(lambda x, y: x + y, number_of_words)
    print(total_number_of_words)

    strings_list = list(map(str.strip, strings))
    strings_list = list(map(lambda x: x.replace(" ", ''), strings_list))
    number_of_string = list(map(lambda x: len(x), strings_list))
    total_number_of_characters = reduce(lambda x, y: x + y, number_of_string)
    print(total_number_of_characters)
```

```python
def change_case(func):
    def myinner():
        return func().upper()
    return myinner

@change_case
def myfunction():
    return "hello bok"
print(myfunction())
```

```python
def count_down(n):
    if n <= 0:
        print("Done!")
    else:
        print(n)
        count_down(n-1)

count_down(5)
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
print(factorial(6))
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)
print(fibonacci(7))
```

```python
def sum_list(numbers):
    if len(numbers) == 0:
        return 0
    else:
        return numbers[0] + sum_list(numbers[1:])

my_list =[1,2,3,4,5]
print(sum_list(my_list))
```
