# fibonacci using for loop

```python
n = int(input("please enter more than 2: "))
first_fib = 0
second_fib = 1
new_fib = 0

for i in range(n-2):
    new_fib = second_fib + first_fib
    first_fib = second_fib
    second_fib = new_fib
print('The 20th Fibonacci is :', new_fib)
```

# using function

```python
count = 2
def fibonacci (fib1, fib2):
    global count
    if count <= 19:
        new_fib = fib1 + fib2
        print(new_fib)
        fib1 = fib2
        fib2 = new_fib
        count += 1
        fibonacci(fib1, fib2)
    else:
        return

fibonacci(0, 1)
```

# Replace MinMax

```python
n = int(input())
num_list = input()
num = num_list.split()
max =num[0]
min =num[0]

for i in range(n):
    if int(max) < int(num[i]):
        max =  num[i]
    if int(min) > int(num[i]):
        min =num[i]
for i in range(n):
    if num[i] == max:
        num[i] = min
    elif num[i] == min:
        num[i] = max
print(*num)
```

```python
def square_addition(*args):
    summation = 0
    for i in args:
        summation += i**2
    return summation
ans = square_addition(10,20,30,40,50)
print(ans)
```
