# fibonacci using for loop

```python
n = int(input("please enter more than 2: "))
first_fib = 0
second_fib = 1
new_fib = 0

for i in range(n-2):
    new_fib = second_fib + first_fib
    first_fib = second_fib
    second_fib = new_fib
print('The 20th Fibonacci is :', new_fib)
```

```python
n = int(input())
num_list = input()
num = num_list.split()
max =num[0]
min =num[0]

for i in range(n):
    if int(max) < int(num[i]):
        max =  num[i]
    if int(min) > int(num[i]):
        min =num[i]
for i in range(n):
    if num[i] == max:
        num[i] = min
    elif num[i] == min:
        num[i] = max
print(*num)
```

```python
def insertion_sort(arr):
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1]= key
    return arr

num = [12,3,42,5,122,7]
sorted_num = insertion_sort(num)
print(sorted_num)
```
