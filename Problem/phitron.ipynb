Given a list of integers, extract all numbers that are
 divisible by 4 but not divisible by 8.

Input Format:
Space-separated integers

Output Format:
A list of numbers satisfying the condition

Example:
Input:
4 6 8 12 16 20 24 28 

Output:
[4, 12, 20, 28] 

```python
numbers = input().split()
divided_by_4 = []
length = len(numbers)
for i in range(length):
    if int(numbers[i]) % 4 == 0 and int(numbers[i]) % 8 !=0:
        divided_by_4.append(int(numbers[i]))
print(divided_by_4)
```

Write a function compute_ratio(x, y) that returns 
(x + y) / (x * y). Use exception handling to manage
division by zero.


Use exception handling to manage:
Division by zero

Input Format:
Two values

Output Format:
Computed result OR error message

```python
def compute_ratio(x, y):
    return ( x + y) / ( x * y )
try:
    numbers = input().split()
    x = int(numbers[0])
    y = int(numbers[-1])
    print(compute_ratio(x,y))
except:
    print("Division not possible")
```

Write a Python program that reads a text file and finds the longest word.

Requirements:

Open and read data from a file (Example: data.txt)
If multiple words have the same longest length, print any one of them


Input Format:
A text file containing multiple lines of text

Output Format:
The longest word

Example File Content (data.txt):
Python is a wonderful language for machine learning.

Output:
wonderful

```python
with open('./files/data.txt', 'r') as file:
    strings_list =file.readlines()
    strings_list = list(map(lambda x: x.split(), strings_list))
    strings_list = strings_list[0]
    strings_length = list(map(lambda x: len(x), strings_list))
    longest_word = []
    max_word = max(strings_length)
    length_of_strings = len(strings_list)
    for i in range(length_of_strings):
            if max_word == strings_length[i]:
                longest_word.append(strings_list[i])

    print(longest_word[0])
```

Write a program that converts a sentence into a list of words but only includes words that start with a vowel (a, e, i, o, u), ignoring case.

Input Format:

A single line sentence

OutputFormat:
A list of strings

Example:
Input:
An apple a day keeps the doctor away
Output:
['An', 'apple', 'a', 'away']


```python
string = input().split()
string_length = len(string)
list_of_words = []
for i in range(string_length):
    word = list(string[i].lower())
    if ('a' or 'e' or 'i' or 'o' or 'u') in word[0]:
        list_of_words.append(string[i])
print(list_of_words)
```

A word is called a Palindrome if reversing its letters results in the same word (ignoring case). Write a program to check whether a given string is a palindrome.

Input Format:
A single word string

Output Format:
"Palindrome" or "Not Palindrome"

Example:
Input:
Radar

Output:
Palindrome


```python
string = input()
length_of_string = len(string)
word = list(string.lower())
reversed_string = []
for i in range(length_of_string):
    reversed_string.append(word[-(i+1)])
if word == reversed_string:
    print('Palindrome')
else:
    print('Not Palindrome')
      

```
